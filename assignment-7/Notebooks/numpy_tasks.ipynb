```python
import numpy as np

arr = np.arange(1, 11)
print("1D array:", arr)

matrix = np.random.randint(1, 10, (3, 3))
print("2D array:\n", matrix)

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Max:", np.max(arr))

arr2 = arr * 2
print("Multiplied by 2:", arr2)

arr[arr > 5] = 0
print("Replaced >5 with 0:", arr)
```
